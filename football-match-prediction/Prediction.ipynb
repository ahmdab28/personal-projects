```python
def rolling_averages(group, cols, new_cols):
    group=group.sort_values("date")
    rolling_stats = group[cols].rolling(3, closed='left').mean()
    group[new_cols]=rolling_stats
    group=group.dropna(subset=new_cols)
    return group
```

```python
cols=["gf","ga","sh","sot","dist","fk","pk","pkatt"]
new_cols=[f"{c}_rolling" for c in cols]
```

```python
def make_predictions(data,predictors):
    train = data[data["date"]< '2022-01-01']
    test = data[data["date"]> '2022-01-01']
    rf.fit(train[predictors],train["target"])
    preds = rf.predict(test[predictors])
    combined = pd.DataFrame(dict(actual=test["target"],predicted=preds), index=test.index)
    precision=precision_score(test["target"],preds)
    return combined, precision
```

```python
def predict_arsenal_vs(opponent_name, venue="Home", match_hour=15):
    """
    Predict Arsenal's match against a specific opponent
    
    Parameters:
    - opponent_name: Name of the opponent team
    - venue: "Home" or "Away"
    - match_hour: Hour of the match (24-hour format)
    """
    
    # Get Arsenal's recent matches
    arsenal_matches = matches[matches["team"] == "Arsenal"].copy()
    arsenal_matches = arsenal_matches.sort_values("date", ascending=False)
    
    # Get opponent code
    try:
        opp_code = matches[matches["opponent"] == opponent_name]["opp_code"].iloc[0]
    except:
        print(f"Opponent '{opponent_name}' not found in data")
        return None
    
    # Get recent stats for Arsenal (last 3 matches)
    arsenal_recent = arsenal_matches.head(3)
    
    # Calculate rolling averages
    gf_rolling = arsenal_recent["gf"].mean()
    ga_rolling = arsenal_recent["ga"].mean()
    sh_rolling = arsenal_recent["sh"].mean()
    sot_rolling = arsenal_recent["sot"].mean()
    dist_rolling = arsenal_recent["dist"].mean()
    fk_rolling = arsenal_recent["fk"].mean()
    pk_rolling = arsenal_recent["pk"].mean()
    pkatt_rolling = arsenal_recent["pkatt"].mean()
    
    # Create prediction data
    prediction_data = {
        'venue_code': 1 if venue == "Home" else 0,
        'opp_code': opp_code,
        'hour': match_hour,
        'day_code': 5,  # Saturday
        'gf_rolling': gf_rolling,
        'ga_rolling': ga_rolling,
        'sh_rolling': sh_rolling,
        'sot_rolling': sot_rolling,
        'dist_rolling': dist_rolling,
        'fk_rolling': fk_rolling,
        'pk_rolling': pk_rolling,
        'pkatt_rolling': pkatt_rolling
    }
    
    # Convert to DataFrame
    pred_df = pd.DataFrame([prediction_data])
    
    # Make prediction
    predictors = ["venue_code", "opp_code", "hour", "day_code"] + new_cols
    win_probability = rf.predict_proba(pred_df[predictors])[0]
    prediction = rf.predict(pred_df[predictors])[0]
    
    # Display results
    print("=" * 60)
    print(f"ARSENAL vs {opponent_name.upper()}")
    print("=" * 60)
    print(f"Venue: {venue}")
    print(f"Time: {match_hour}:00")
    print("\n📊 PREDICTION:")
    print(f"  Win Probability: {win_probability[1]*100:.1f}%")
    print(f"  Loss/Draw Probability: {win_probability[0]*100:.1f}%")
    
    if prediction == 1:
        print(f"\n🎯 VERDICT: ARSENAL WIN")
    else:
        print(f"\n⚪ VERDICT: ARSENAL NOT WIN")
    
    # Recent form
    print("\n" + "-" * 40)
    print("Arsenal's Recent Form (Last 3 matches):")
    for i, row in arsenal_recent.iterrows():
        result_symbol = {"W": "✓", "L": "✗", "D": "="}[row["result"]]
        print(f"  {result_symbol} vs {row['opponent']} ({row['venue']}) {row['gf']}-{row['ga']}")
    
    return {
        'opponent': opponent_name,
        'venue': venue,
        'win_probability': win_probability[1],
        'prediction': 'Win' if prediction == 1 else 'Not Win'
    }

# Example usage:
# Predict Arsenal vs Tottenham at home
prediction = predict_arsenal_vs("Tottenham", venue="Home", match_hour=12)

# Predict Arsenal vs Manchester City away
prediction2 = predict_arsenal_vs("Manchester City", venue="Away", match_hour=16)
```
